```python
from collections import deque
import heapq
```

```python
# Graph Representation

graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 2, 'E': 5},
    'C': {'F': 3},
    'D': {},
    'E': {'F': 1},
    'F': {}
}

# Heuristic Values

heuristic = {
    'A': 6,
    'B': 4,
    'C': 2,
    'D': 5,
    'E': 1,
    'F': 0
}

print("Graph Loaded Successfully")
```

```python
def explain(message):
    print("[EXPLAIN]:", message)
```

```python
def bfs(start, goal):

    explain("Starting BFS Search")

    queue = deque([[start]])
    visited = set()

    while queue:

        path = queue.popleft()
        node = path[-1]

        explain(f"Visiting Node: {node}")

        if node == goal:
            explain("Goal Found")
            return path

        if node not in visited:

            visited.add(node)

            for neighbor in graph[node]:

                new_path = list(path)
                new_path.append(neighbor)

                queue.append(new_path)

    return None
```

```python
def dfs(start, goal):

    explain("Starting DFS Search")

    stack = [[start]]
    visited = set()

    while stack:

        path = stack.pop()
        node = path[-1]

        explain(f"Visiting Node: {node}")

        if node == goal:
            explain("Goal Found")
            return path

        if node not in visited:

            visited.add(node)

            for neighbor in graph[node]:

                new_path = list(path)
                new_path.append(neighbor)

                stack.append(new_path)

    return None
```

```python
def astar(start, goal):

    explain("Starting A* Search")

    pq = []

    heapq.heappush(pq, (0, [start]))

    visited = set()

    while pq:

        cost, path = heapq.heappop(pq)

        node = path[-1]

        explain(f"Expanding Node: {node}")

        if node == goal:
            explain("Goal Reached")
            return path

        if node not in visited:

            visited.add(node)

            for neighbor in graph[node]:

                g = cost + graph[node][neighbor]
                h = heuristic[neighbor]
                f = g + h

                explain(f"{neighbor} -> g={g}, h={h}, f={f}")

                new_path = list(path)
                new_path.append(neighbor)

                heapq.heappush(pq, (f, new_path))

    return None
```

```python
colors = ['Red', 'Green', 'Blue']

regions = {
    'WA': ['NT', 'SA'],
    'NT': ['WA', 'SA', 'Q'],
    'SA': ['WA', 'NT', 'Q', 'NSW', 'V'],
    'Q': ['NT', 'SA', 'NSW'],
    'NSW': ['Q', 'SA', 'V'],
    'V': ['SA', 'NSW']
}
```

```python
def valid(region, color, assignment):

    for neighbor in regions[region]:

        if neighbor in assignment and assignment[neighbor] == color:
            return False

    return True


def solve(assignment={}):

    if len(assignment) == len(regions):
        return assignment

    unassigned = [r for r in regions if r not in assignment]

    region = unassigned[0]

    for color in colors:

        explain(f"Trying {color} for {region}")

        if valid(region, color, assignment):

            assignment[region] = color

            result = solve(assignment)

            if result:
                return result

            explain(f"Backtracking from {region}")

            del assignment[region]

    return None
```

```python
print("\n========== BFS RESULT ==========")

result_bfs = bfs('A', 'F')

print("BFS Path:", result_bfs)
```

```python
print("\n========== DFS RESULT ==========")

result_dfs = dfs('A', 'F')

print("DFS Path:", result_dfs)
```

```python
print("\n========== A* RESULT ==========")

result_astar = astar('A', 'F')

print("A* Path:", result_astar)
```

```python
print("\n========== CSP RESULT ==========")

result_csp = solve()

print("Color Assignment:", result_csp)
```
